# EXPERIMENT 10 - K MEANS CLUSTERING

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns 
import sklearn 
from sklearn.cluster import KMeans
```

```python
sns.set()
```

```python
# Load the data
data = pd.read_csv('3.01.Countryclusters.csv')
data
```

```python
# Plotting the dataset
plt.scatter(data['Latitude'],data['Longitude'])
plt.xlim(-180,180)
plt.ylim(-90,90)
plt.show()
```

```python
# SELECTING THE FEATURES
x = data.iloc[:,1:3]
x
```

```python
# Clustering 
kmeans = KMeans(5)
# fitting the data present in X in kmean
kmeans.fit(x)
# Clustering Result
# fitting the data into the model
identified_clusters = kmeans.fit_predict(x)
identified_clusters
# creating a copy of our data 
data_with_clusters = data.copy()
# new cluster--> data
data_with_clusters['Clusters'] = identified_clusters
plt.scatter(data_with_clusters['Latitude'],data_with_clusters['Longitude'],c = data_with_clusters['Clusters'],cmap='rainbow')
plt.show()
```

```python
import math
data = [(10, 2), (12, 4), (16, 6), (20, 8),
        (24, 10), (26, 12), (30, 14), (34, 16)]
c1 = data[0]
c2 = data[-1]
while True:
    cluster1 = []
    cluster2 = []
    for p in data:
        d1 = math.sqrt((p[0] - c1[0]) ** 2 + (p[1] - c1[1]) ** 2)
        d2 = math.sqrt((p[0] - c2[0]) ** 2 + (p[1] - c2[1]) ** 2)
        if d1 < d2:
            cluster1.append(p)
        else:
            cluster2.append(p)
    x1 = 0
    y1 = 0
    for p in cluster1:
        x1 = x1 + p[0]
        y1 = y1 + p[1]
    x1 = x1 / len(cluster1)
    y1 = y1 / len(cluster1)
    new_c1 = (x1, y1)
    x2 = 0
    y2 = 0
    for p in cluster2:
        x2 = x2 + p[0]
        y2 = y2 + p[1]
    x2 = x2 / len(cluster2)
    y2 = y2 / len(cluster2)
    new_c2 = (x2, y2)
    print("Cluster 1:", cluster1)
    print("Cluster 2:", cluster2)
    print("Centroid 1:", new_c1)
    print("Centroid 2:", new_c2)
    print()
    if new_c1 == c1 and new_c2 == c2:
        break
    c1 = new_c1
    c2 = new_c2
print("Final Clusters:")
print("Cluster 1 =", cluster1)
print("Cluster 2 =", cluster2)
print("Centroid 1 =", c1)
print("Centroid 2 =", c2)
```
